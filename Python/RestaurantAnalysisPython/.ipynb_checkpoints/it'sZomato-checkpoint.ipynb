### Importing libraries:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
df = pd.read_csv("C:\\Users\\Pitruja\\OneDrive\\Documents\\Zomato Project\\zomatoData.csv")
df.head()
```

```python
def handleRating(values):
    values = str(values).split('/')
    values = values[0]
    return float(values)
```

```python
df['rate'] = df['rate'].apply(handleRating)
df.head()
```

```python
df.info()
```

### Types of Restaurant:

```python
df.head()
```

```python
sns.countplot(x = df['listed_in(type)'])
plt.xlabel('Types of restaurant:')
```

### Number of votes vs type of restaurant:

```python
grouped_data = df.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c = 'green', marker = 'o')
plt.xlabel('Types of restaurant', fontsize = 20)
plt.ylabel('Votes', fontsize = 20)
```

### Check what rating is maximum:

```python
df.head()
```

```python
plt.hist(df['rate'], bins = 5)
plt.title('Rating distribution')
plt.show()
```

### Average order spending by couples:

```python
df.head()
```

```python
couple_data = df['approx_cost(for two people)']
sns.countplot(x = couple_data)
```

### Which mode recieves maximum ratings(online or offline):

```python
df.head()
```

```python
plt.figure(figsize = (6,6))
sns.boxplot(x = 'online_order', y = 'rate', data = df)
```

### Types of restaurant which order online or offline:

```python
pivot_table = df.pivot_table(index = 'listed_in(type)', columns = 'online_order', aggfunc = 'size', fill_value = 0)
sns.heatmap(pivot_table, annot = True, cmap = 'YlGnBu', fmt = 'd')
plt.title('Heatmap')
plt.xlabel('Online Order')
plt.ylabel('Listed in (Type)')
plt.show()
```
